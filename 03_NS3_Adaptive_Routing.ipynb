# Make a separate copy of the NS-3 simulation

```python
from pathlib import Path
import shutil

source_file = Path("ns3/topology.cc")
new_file = Path("ns3/adaptive_topology.cc")

if not source_file.is_file():
    raise FileNotFoundError(f"Cannot find {source_file}")

if new_file.exists():
    print("Separate simulation file already exists:", new_file)
else:
    shutil.copy2(source_file, new_file)
    print("Created separate simulation file:", new_file)

print("Original topology preserved:", source_file.is_file())
```

```python
lines = cpp_code.splitlines()

for keyword in [
    "Ipv4GlobalRoutingHelper",
    "PopulateRoutingTables",
    "InternetStackHelper",
    "NodeContainer",
    "Ipv4AddressHelper"
]:
    print(f"\n{'=' * 15} {keyword} {'=' * 15}")

    matches = [
        i for i, line in enumerate(lines)
        if keyword in line
    ]

    if not matches:
        print("Not found")
        continue

    for i in matches:
        start = max(0, i - 5)
        end = min(len(lines), i + 8)

        for j in range(start, end):
            print(f"{j + 1}: {lines[j]}")
        print()
```
