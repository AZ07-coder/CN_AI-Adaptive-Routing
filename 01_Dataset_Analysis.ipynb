## Load our existing dataset

```python

import csv
from pathlib import Path

dataset_file = Path("dataset/processed/flow_analysis.csv")

if dataset_file.exists():
    with open(dataset_file, "r", newline="") as file:
        rows = list(csv.DictReader(file))

    print("Dataset loaded successfully!")
    print("Total flow records:", len(rows))
    print("Columns:", list(rows[0].keys()) if rows else "No rows")
else:
    print("Dataset not found:", dataset_file.resolve())
```
